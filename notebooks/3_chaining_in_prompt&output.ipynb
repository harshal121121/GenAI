**Chaining In Prompts**

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm=ChatOpenAI(model="gpt-5-mini")
output=StrOutputParser()
#output parser is used to parse the output of llm and convert it into string format. we dont need to add .content in last to get content part of llm output. it will automatically convert the output into string format.

prompts=ChatPromptTemplate.from_messages([

    {"role":"system", "content":"You are a Translator and convert input into {language}."},
    {"role":"user", "content":"{query}" }
])

#here we have created dynamic prompting which is taking language and query as a input variables. 


chains= prompts | llm | output
#here we have created a chaining in which prompts output goes to llm as a input then llm output goes to output parser as a input. then further we can also add multiple chains whatever operations we have to do on output like upper case or lower case string only whatever.

res=chains.invoke({"language":"Marathi", "query":"I love Langchain."})
print(res)

#llm was runnable then prompts was also runnable then output is also becomes runnable and chains also becomes runnable automatically. so we can invoke the chains directly with input variables.
```

```python
def upper_case(input_string):
    return input_string.upper()

chains= prompts | llm | output | upper_case
res=chains.invoke({"language":"Hinglish", "query":"I love Langchain."})
print(res)

#here we have performed another operation on output string which is upper case , like this we can perform multiple operation and can use chaning to create complex operations on output  and can also send on api. that's why chaining is very useful in Langchain. 
```
