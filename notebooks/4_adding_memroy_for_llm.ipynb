```python
from dotenv import load_dotenv
load_dotenv()
from langchain_openai import ChatOpenAI
llm=ChatOpenAI(model_name="gpt-5-mini")
```

**Managing History for llm**

```python
history=[]
while True:
    query=input("User: ")
    if query.lower() in ["exit", "quit", "bye"]:
        print("Exiting the chat. Goodbye!👋")
        break
    history.append({"role": "user", "content": query})
    print("User:", query)
    res=llm.invoke(history)
    history.append({"role": "ai", "content": res.content})
    print("AI:", res.content, "\n")

#here we have created list of history which we are passing to next prompt. we are passing user query and llm response of that to next one and also printing both of them. This way we can have a conversation with the llm and it will remember the previous queries and responses.
```
