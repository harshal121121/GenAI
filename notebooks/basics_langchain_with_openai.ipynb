**Static Prompting**

```python
from langchain_openai import ChatOpenAI
llm=ChatOpenAI(model_name="gpt-5-mini")
prompts=[
    ("system", "You are Python developer."),
    ("user", "Write a only code to add two numbers."),
]

res=llm.invoke(prompts)
print(res.content)

# This is static prompting where we have passed list of tuples as prompts.in which we can pass type nd prompt for tuple. there are 3 types system, user and ai , so llm will get the knowledge for whom that prompt is for.
```

**Langchain using Anthropic Model**

```python
from langchain_anthropic import ChatAnthropic
llm=ChatAnthropic(model="claude-haiku-5-5")
prompts=[
    ("system", "You are Java developer."),
    ("user", "Write a only code of binary search")
]
res=llm.invoke(prompts)
print(res.content)
```

**Dynamic Prompting**

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
llm=ChatOpenAI(model_name="gpt-5-mini")

prompts=ChatPromptTemplate.from_messages([
    {"role":"system", "content":"You are {llm_role} and convert input into {language}"},
    {"role":"user", "content":"{query}"}
])

new_prompt=prompts.format_messages(llm_role="Translator", language="Marathi", query="I love you so much.")
res=llm.invoke(new_prompt)
print(res.content)
```
